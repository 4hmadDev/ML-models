```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import accuracy_score
```

DATA SET :

```python
df = pd.read_csv("/content/drive/MyDrive/datasets/coin_Dogecoin.csv")
```

 Droping coloums which are not needed :

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
X = df.drop(columns=['Open','Date','Volume','Marketcap','Close','Low'])
```

Creating Dependent and Independent Variables :

```python
X = df[['Open', 'High', 'Low', 'Volume', 'Marketcap']]
Y = df['Close']
```

for traing and finding the accuracy of the dataset :

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

using decision tree model and finding accuracy in the dataset :

```python
model = DecisionTreeRegressor(random_state=42)
model.fit(X_train, Y_train)
```

```python
model.predict(X_test)
```

```python
accuracy = model.score(X_test, Y_test)
print("Accuracy:", accuracy)
```
