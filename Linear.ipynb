```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

DATA SET :

```python
df = pd.read_csv("/coin_Dogecoin.csv")
```

Creating Dependent and Independent Variables :

```python
X = df[['High','Marketcap','Low','Open']]
Y = df['Close']
```

```python
X = df.drop(columns=['SNo', 'Name', 'Symbol', 'Date'])
```

for traing and finding the accuracy of the dataset :

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

using linear regression model and finding accuracy in the dataset :

```python
model = LinearRegression()
model.fit(X_train, Y_train)
```

```python
Y_pred = model.predict(X_test)
r2 = r2_score(Y_test, Y_pred)

print(f"Accuracy: {r2}")
```
